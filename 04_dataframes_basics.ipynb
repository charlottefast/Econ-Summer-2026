```python
# Prediction: boolean mask of body mass > 4500, equal rows to original
df["body_mass_g"] > 4500
```

```python
# Prediction: rows where body mass > 4500, fewer rows than original
df[df["body_mass_g"] > 4500]
```

```python
df[(df["body_mass_g"] > 4500) & (df["island"] == "Dream")]
```

```python
# Objective: filter rows where sex is female
df[df["sex"] == "FEMALE"]
```

```python
# Objective: filter rows where body mass is < 3000 or > 5500
df[(df["body_mass_g"] > 5500) | (df["body_mass_g"] < 3000)]
```

```python
df.sort_values("body_mass_g")
```

```python
# Objective: return table of five heaviest penguins
df.sort_values("body_mass_g", ascending=False).head(5)
```

Question: Which island's penguins are heaviest on average?

```python
# Description: sort the data by island and return a table of each island's mean body mass

island_means = df.groupby("island")["body_mass_g"].mean()
island_means.idxmax()
```
