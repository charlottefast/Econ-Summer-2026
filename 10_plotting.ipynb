```python
import pandas as pd
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/penguins.csv"
penguins = pd.read_csv(url)
penguins
```

```python
penguins.groupby("species")["body_mass_g"].mean().plot(kind="bar")
```

```python
penguins.groupby("species")["body_mass_g"].mean().plot(kind="barh")
```

```python
import matplotlib.pyplot as plt

penguins.groupby("species")["body_mass_g"].mean().plot(kind="bar")
plt.title("Mean Body Mass by Penguin Species")
plt.xlabel("Species")
plt.ylabel("Mean body mass (g)")
plt.show()
```

```python
penguins.plot(kind="scatter", x="bill_length_mm", y="body_mass_g")
plt.title("Body Mass vs Bill Length")
plt.xlabel("Bill length (mm)")
plt.ylabel("Body mass (g)")
plt.show()
```

```python
penguins.plot(kind="scatter", x="bill_length_mm", y="body_mass_g", c="species", cmap="viridis")
```

```python
penguins["body_mass_g"].plot(kind="hist", bins=20)
plt.title("Distribution of Body Mass")
plt.xlabel("Body mass (g)")
plt.show()
```

```python
# Objective: a bar chart with mean flipper length by island, fully labeled
penguins.groupby("island")["flipper_length_mm"].mean().plot(kind="bar")
```

```python
# Objective: a scatter plot with flipper length vs body mass, labeled, colored by species
penguins.plot(kind="scatter", x="flipper_length_mm", y="body_mass_g", c="species", cmap="viridis")
```

```python
# Objective: a histogram showing a distribution of bill depth with sensible bin count
penguins["bill_depth_mm"].plot(kind="hist", bins=20)
```

Notes:
- Bar charts are effective for comparing categorical data across groups
- Scatter plots are useful for visualizing relationships between two continuous variables
- Histograms are ideal for showing the distributio nof a single continuous variable
