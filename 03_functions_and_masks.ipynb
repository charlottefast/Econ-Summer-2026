### Part 1: Functions

```python
def double(x):
    return x * 2
```

```python
# prediction: 10
double(5)
```

```python
# prediction: error
double("CS")
```

```python
courses = ["CS106A", "CS106B", "MATH51", "CS109"]
```

```python
def first_and_last(items):
    return items[0], items[-1]
```

```python
#prediction: ("CS106A", "CS109")
first_and_last(courses)
```

```python
# prediction: 20
result = double(10)
print(result)
```

```python
# prediction: "hello!"
def shout(word):
    print(word + "!")

result = shout("hello")
print(result)
```

```python
def letter_grade_to_points(grade):
    if grade == "A":
        return 4.0
    elif grade == "A-":
        return 3.7
    elif grade == "B+":
        return 3.3
    else:
        return 0.0
```

```python
# predictions: 4.0, 3.7, 3.3, 0.0
print(letter_grade_to_points("A"))
print(letter_grade_to_points("A-"))
print(letter_grade_to_points("B+"))
print(letter_grade_to_points("C"))
```

```python
letter_grades = ["A", "A-", "B+"]
```

```python
# prediction: [4.0, 3.7, 3.3]
[letter_grade_to_points(grade) for grade in letter_grades]
```

### Part 2: Boolean Masks

```python
import numpy as np
wages = np.array([45000, 82000, 31000, 120000, 67000])
```

```python
# prediction: [90000, 164000, 62000, 240000, 134000]
wages * 2
```

```python
# prediction: [45, 82, 31, 120, 67]
wages / 1000
```

```python
[45000, 82000] * 2
```

```python
# Prediction: [82000, 120000, 67000]
wages > 60000
```

```python
# Prediction: [82000, 120000, 67000]
wages[wages > 60000]
```

```python
# Prediction: []
wages[wages > 200000]
```

```python
# Prediction: 3
(wages > 60000).sum()
```

```python
# Prediction: 0.6
(wages > 60000).mean()
```
