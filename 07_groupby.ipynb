```python
import pandas as pd
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/penguins.csv"
penguins = pd.read_csv(url)
penguins
```

```python
(penguins.groupby("island")["body_mass_g"].mean()).idxmax()
```

```python
penguins["species"].value_counts()
```

```python
penguins.groupby("species")["body_mass_g"].mean()
```

```python
penguins.groupby("species")["flipper_length_mm"].max()
```

```python
penguins.groupby("species").count()
```

```python
penguins.groupby("species").size()
```

```python
penguins.groupby("species")["body_mass_g"].agg(["mean", "min", "max", "count", "median"])
```

```python
penguins.groupby(["species", "island"])["body_mass_g"].mean()
```

```python
penguins.groupby("species")["body_mass_g"].mean().sort_values(ascending=False)
```

```python
penguins.groupby("species")["body_mass_g"].mean().sort_values(ascending=False).idxmax()
```

```python
penguins.groupby("species")["body_mass_g"].mean().sort_values(ascending=False).head(1)
```

Question: Which species has the longest average bill?

```python
penguins.groupby("species")["bill_length_mm"].mean().sort_values(ascending=False).head(1)
```

Question: On each island, how many penguins of each species?

```python
penguins.groupby(["island", "species"]).size()
```

Question: Among female penguins only, what is the mean body mass by species?

```python
penguins[penguins["sex"] == "FEMALE"].groupby("species")["body_mass_g"].mean()
```

Notes
- `.groupby()` splits a table into groups based on a certain column's values, computes something within each group, and returns a table with one row per group
- `.size()` counts the number of rows, `.count()` counts the number of values within each column of a row
