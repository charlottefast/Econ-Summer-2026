### Data cleaning notes - mystery_bls.csv, 8 rows x 5 columns as received
- Column names contained stray whitespace (e.g. ' median_wage'); stripped
- Occupation titles contained leading/trailing whitespace in 3 of 8 rows; stripped
- `median_wage` and `total_employment` loaded as text because missing values were coded as text; coerced to numeric with '**' -> NaN
- Duplicate row (Software Developers, 15-1252) appeared twice; deduplicated, 8 -> 7 rows
- After coercion, 2 of 7 rows had one missing value each (Data Scientists: employment; Elementary Teachers: wage); **decided** to drop both, 7 -> 5 rows, to work with complete cases; cost: analysis loses one large high-wage occupation and one large public-sector one; alternative: retain rows and use NaN-aware operations
